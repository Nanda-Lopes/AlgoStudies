# Problema da Mochila (Instância Menor)

O problema da mochila consiste em selecionar um subconjunto de itens com pesos e valores associados de forma que o valor total seja maximizado e a soma dos pesos não exceda a capacidade máxima $W$ da mochila.

Para instâncias de tamanho moderado, a abordagem de Programação Dinâmica iterativa (*bottom-up*) com otimização de espaço foi empregada:
1. Uma tabela unidimensional `dp` de tamanho $W + 1$ foi inicializada com zeros.
2. Cada item foi processado sequencialmente de $1$ a $n$.
3. Para cada item, a tabela de estados foi atualizada de trás para frente (de $W$ até o peso do item atual) para evitar que o mesmo item fosse computado mais de uma vez na mesma etapa.
4. O valor final armazenado em `dp[W]` representa a solução ótima para o limite de peso estabelecido.

```python
def solve_knapsack_small(filename):
    with open(filename, "r") as f:
        header = f.readline()
        capacity, num_items = map(int, header.split())
        items = []
        for line in f:
            if line.strip():
                val, wt = map(int, line.split())
                items.append((val, wt))
    dp = [0] * (capacity + 1)
    for val, wt in items:
        for w in range(capacity, wt - 1, -1):
            if dp[w - wt] + val > dp[w]:
                dp[w] = dp[w - wt] + val
    return dp[capacity]

result_small = solve_knapsack_small("knapsack1.txt")
print("==================================")
print("RESPOSTA QUESTÃO 1 (OPTIMAL VALUE):")
print(result_small)
print("==================================")
```

# Problema da Mochila (Instância Massiva)

A segunda instância possui uma capacidade de mochila grande ($W = 2.000.000$) e uma quantidade maior de itens ($n = 2.000$). Uma tabela bidimensional de tamanho $2000 \times 2000000$ consumiria uma quantidade inviável de memória, estourando os limites padrão do sistema.

Para contornar essa restrição física, uma estratégia de Programação Dinâmica recursiva com memorização (*top-down*) foi implementada:
1. Um dicionário de buscas (`cache`) foi definido para armazenar apenas as soluções dos subproblemas que são efetivamente necessários durante a execução.
2. A árvore de decisão recursiva inicia no último item e na capacidade total $W$.
3. Em cada estado $(i, w)$, se o peso do item atual for superior à capacidade disponível $w$, o item foi obrigatoriamente descartado, e o cálculo prossegue para o estado anterior $(i-1, w)$.
4. Caso contrário, o algoritmo calcula o valor máximo entre duas escolhas: descartar o item ou incluí-lo na mochila (reduzindo a capacidade disponível pelo peso do item e adicionando o seu valor).
5. O resultado de cada estado foi armazenado no dicionário antes de ser retornado, prevenindo a recomputação de ramos repetidos na árvore recursiva e garantindo a otimização de espaço em memória.

```python
import sys
sys.setrecursionlimit(200000)

def solve_knapsack_big(filename):
    with open(filename, "r") as f:
        header = f.readline()
        capacity, num_items = map(int, header.split())
        items = []
        for line in f:
            if line.strip():
                val, wt = map(int, line.split())
                items.append((val, wt))
    memo = {}
    def knapsack_recursive(i, w):
        if i < 0 or w <= 0:
            return 0
        state = (i, w)
        if state in memo:
            return memo[state]
        val, wt = items[i]
        if wt > w:
            res = knapsack_recursive(i - 1, w)
        else:
            res = max(
                knapsack_recursive(i - 1, w),
                knapsack_recursive(i - 1, w - wt) + val
            )
        memo[state] = res
        return res
    return knapsack_recursive(num_items - 1, capacity)

result_big = solve_knapsack_big("knapsack_big.txt")
print("==================================")
print("RESPOSTA QUESTÃO 2 (OPTIMAL VALUE):")
print(result_big)
print("==================================")
```
