# Median Maintenance
Implementação do algoritmo de Manutenção de Mediana para processamento de fluxo contínuo de 10.000 números.

* Cálculo da mediana inferior a cada novo número inserido
* Finalizando com a soma de todas as 10.000 medianas geradas para extrair e retornar o módulo 10000 dessa soma.

Dois Heaps são utilizados para maximizar a solução:
* um Max-Heap para a metade inferior dos números e
* um Min-Heap para a metade superior

## Leitura do arquivo e estrutura base
1. Importação da biblioteca 'heapq', usada para gerenciar os dois heaps, já contornando o padrão Min-Heaps do Python.
2. Criação de uma lista vazia 'stream'.
3. Leitura do arquivo 'Median.txt' linha por linha, convertendo as strings em números inteiros e os adicionando sequencialmente na lista.

```python
import heapq

filename = 'Median.txt'
stream = []

with open(filename, 'r') as f:
    for line in f:
        if line.strip():
            stream.append(int(line.strip()))
```

## Algoritmo de Manutenção de Mediana
1. Criação de duas listas vazias:
   * 'low_heap' (Max-Heap para os menores números)
   * 'high_heap' (Min-Heap para os maiores).
2. Busca em loop que percorre a lista 'stream'. Note que o novo número é inserido no 'low_heap' caso seja menor ou igual à raiz atual do 'low_heap',ou ainda se o heap estiver vazio. Caso contrário, segue para o 'high_heap'.
3. Verificação de equilibrio, balanceando a divisão equitativa entre o 'low_heap' e o 'high_heap', permitindo no máximo 1 elemento a mais.
4. Extração da mediana, acompanhando a regra de balanceamento restrita, deixando a mediana do momento (mediana inferior) equivalente à raiz do 'low_heap'.

```python
def median_maintenance(stream):
    low_heap = []
    high_heap = []
    medians = []

    for x in stream:
        # Fix: Access the root of the heap with [0]
        if not low_heap or x <= -low_heap[0]:
            heapq.heappush(low_heap, -x)
        else:
            heapq.heappush(high_heap, x)

        # Rebalance heaps
        if len(low_heap) > len(high_heap) + 1:
            heapq.heappush(high_heap, -heapq.heappop(low_heap))
        elif len(high_heap) > len(low_heap):
            heapq.heappush(low_heap, -heapq.heappop(high_heap))

        # Fix: Access the root of the heap with [0]
        medians.append(-low_heap[0])

    return medians
```

## Execução e formatação da resposta
1. Execução da busca construída na célula anterior considerando o fluxo de números original como entrada.
2. Cálculo da soma de todas as 10.000 medianas acumuladas na lista de resultados
3. Formatação da resposta calculando o módulo 10000 da soma, gerando os 4 últimos dígitos exatos para submissão.


```python
medians = median_maintenance(stream)
total_sum = sum(medians)
final_answer = total_sum % 10000

print("==================================")
print("RESPOSTA FINAL:")
print(final_answer)
print("==================================")
```
