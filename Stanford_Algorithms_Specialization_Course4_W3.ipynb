# Problema do Caixeiro Viajante (TSP) via Heurística do Vizinho Mais Próximo

O Problema do Caixeiro Viajante (TSP) para instâncias massivas torna abordagens exatas computacionalmente intratáveis. Para instâncias com dezenas de milhares de cidades ($n = 33.708$), métodos heurísticos são empregados para obter aproximações viáveis em tempo polinomial.

A **Heurística do Vizinho Mais Próximo (*Nearest Neighbor*)** constrói um circuito iniciando em uma cidade predeterminada e selecionando iterativamente a cidade não visitada mais próxima:

1. O circuito inicia obrigatoriamente na primeira cidade (cidade 1).
2. Em cada etapa, a partir da cidade atual $u$, a próxima cidade $v$ a ser visitada é escolhida de modo a minimizar a distância euclidiana:

$$d(u, v) = \sqrt{(x_u - x_v)^2 + (y_u - y_v)^2}$$

3. Critério de desempate: caso duas ou mais cidades não visitadas apresentem a mesma distância mínima em relação à cidade atual, a cidade de menor índice numérico (identificador) é escolhida.
4. Após visitar todas as $n$ cidades exatamente uma vez, o circuito é fechado calculando a distância de retorno da última cidade visitada até a cidade inicial (Cidade 1).
5. O custo total do circuito corresponde à soma de todos os deslocamentos consecutivos somado à aresta de retorno. O valor final a ser submetido é arredondado para baixo.

```python
import os
import math
import numpy as np

def read_tsp_nn(filename):
    if not os.path.exists(filename) and os.path.exists(filename.replace(".txt", "")):
        filename = filename.replace(".txt", "")
    elif not os.path.exists(filename) and os.path.exists(filename + ".txt"):
        filename = filename + ".txt"

    with open(filename, "r") as f:
        lines = [line.strip() for line in f if line.strip()]
    num_cities = int(lines[0])
    coords = np.empty((num_cities, 2), dtype=np.float64)
    for i, line in enumerate(lines[1:]):
        parts = line.split()
        coords[i, 0] = float(parts[1])
        coords[i, 1] = float(parts[2])
    return num_cities, coords

def solve_tsp_nearest_neighbor(filename):
    num_cities, coords = read_tsp_nn(filename)
    visited = np.zeros(num_cities, dtype=bool)

    current_city = 0
    visited[0] = True
    total_cost = 0.0

    num_visited = 1
    unvisited_indices = np.arange(num_cities, dtype=np.int32)

    while num_visited < num_cities:
        unvisited = unvisited_indices[~visited]
        curr_x = coords[current_city, 0]
        curr_y = coords[current_city, 1]

        dx = coords[unvisited, 0] - curr_x
        dy = coords[unvisited, 1] - curr_y
        dist_sq = dx * dx + dy * dy

        best_pos = np.argmin(dist_sq)
        next_city = unvisited[best_pos]

        total_cost += math.sqrt(dist_sq[best_pos])
        visited[next_city] = True
        current_city = next_city
        num_visited += 1

    total_cost += math.dist(coords[current_city], coords[0])
    return total_cost

tour_cost = solve_tsp_nearest_neighbor("nn.txt")
print("==================================")
print("CUSTO TOTAL DO CIRCUITO:")
print(tour_cost)
print("RESPOSTA OFICIAL (ARREDONDADO PARA BAIXO):")
print(math.floor(tour_cost))
print("==================================")
```
