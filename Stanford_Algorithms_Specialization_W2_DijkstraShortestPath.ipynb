# Dijkstra's Shortest-Path Algorithm
Cálculo do caminho mais curto partindo do vértice 1 para todos os outros em um grafo de 200 vértices, usando o Algoritmo de Dijkstra otimizado com Filas de Prioridade (Heaps).
Vale notar:
* Distâncias inalcançáveis recebem o valor padrão de 1000000
* A saída extrai as distâncias finais para 10 vértices alvos específicos solicitados pelo exercício.

## Leitura do arquivo e montagem do grafo
1. Importação da biblioteca 'heapq', usada para gerenciar filas de prioridade, que é essencial para otimizar o tempo do algoritmo.
2. Criação de um dicionário vazio 'graph', inicializando os vértices de 1 a 200 com listas vazias.
3. Leitura do arquivo linha por linha, extraindo o primeiro número (pop(0)) como vértice de origem.
4. Loop para quebrar o restante da linha, separando os vértices de destino e os pesos das arestas, guardando essas duplas (tuplas) na lista do respectivo vértice de origem.

```python
import heapq

filename = 'dijkstraData.txt'

graph = {}
for i in range(1, 201):
    graph[i] = []

with open(filename, 'r') as f:
    for line in f:
        parts = line.split()
        if not parts:
            continue

        u_str = parts.pop(0)
        u = int(u_str)

        for edge_str in parts:
            v_str, w_str = edge_str.split(',')
            v = int(v_str)
            w = int(w_str)
            graph[u].append((v, w))

print("Grafo carregado com sucesso!")
```

## O Algoritmo de Dijkstra
1. Criação de um dicionário 'distances' com valor padrão 1000000 para todos os vértices, permitindo lidar com nós inalcançáveis (exceto o nó inicial que começa com 0).
2. Criação de fila de prioridade 'pq', iniciando apenas com o nó de origem e de um conjunto 'explored' para rastrear vértices que tiveram caminhos mínimos garantidos.
3. Busca em loop principal, que extrai sempre o vértice mais próximo da fila. Caso já tenha sido explorado, será ignorado, mas uma vez que ainda não tenha sido, passa a ser marcado como tal.
4. Verificação dos vizinhos: caso a nova distância passando pelo vértice atual for menor que a distância registrada anteriormente, a distância é atualizada e o vizinho é empurrado para a fila de prioridade.

```python
def run_dijkstra(graph, start_node):
    distances = {node: 1000000 for node in graph}
    distances[start_node] = 0

    pq = [(0, start_node)]
    explored = set()

    while pq:
        current_dist, current_node = heapq.heappop(pq)

        if current_node in explored:
            continue

        explored.add(current_node)

        for neighbor, weight in graph[current_node]:
            if neighbor in explored:
                continue

            distance = current_dist + weight

            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(pq, (distance, neighbor))

    return distances
```

## Execução e formatação da resposta
1. Execução da busca construída no Bloco 2 partindo do vértice 1.
2. Definição da lista 'targets' com os 10 vértices exigidos pelo exercício (7, 37, 59...).
3. Extração das distâncias finais calculadas para esses alvos.
4. Formatação da resposta, convertendo os números extraídos para string e os unindo com vírgulas, gerando a saída final sem espaços.

```python
distances = run_dijkstra(graph, 1)

targets = [7, 37, 59, 82, 99, 115, 133, 165, 188, 197]
results = []

for t in targets:
    results.append(str(distances[t]))

final_answer = ",".join(results)

print("==================================")
print("RESPOSTA FINAL:")
print(final_answer)
print("==================================")
```
