# 2-SUM Algorithm

Implementação de uma variante do algoritmo 2-SUM para processar um arquivo com 1 milhão de inteiros.
* Cálculo da quantidade de valores-alvo t no intervalo de -10000 a 10000 (inclusive).
* A premissa é que existam dois números distintos no arquivo cuja soma seja igual a t (x+y=t).

Para maximizar a solução e evitar travamento do Colab, haverá:
* Ordenação prévia
* Busca Binária.

## Leitura do arquivo e estrutura base
1. Importação da biblioteca 'bisect', que realizará buscas binárias rápidas e otimizará o tempo de execução no array.
2. Criação de uma lista vazia 'nums' e leitura do arquivo 'algo1-programming_prob-2sum.txt' linha por linha, convertendo as strings em números inteiros.
3. Remoção de elementos duplicados via conversão para um conjunto 'set', e ordenação da lista final, garantindo que os números sejam únicos para a garantia de valores distintos.

```python
import bisect

filename = 'algo1-programming_prob-2sum.txt'
nums = []

with open(filename, 'r') as f:
    for line in f:
        if line.strip():
            nums.append(int(line.strip()))

nums = sorted(list(set(nums)))
```

## Algoritmo 2-SUM Otimizado
1. Criação de um conjunto vazio 'targets', que registrará as somas que satisfazem o intervalo exigido.
2. Busca em loop, que percorrerá a lista ordenada 'nums', tratando cada número como a variável 'x'.
3. Cálculo dinâmico dos limites de busca para 'y', usando as margens para localizar 'y' entre -10000 - x e 10000 - x, garantindo a premissa de que a soma precisa estar entre -10000 e 10000.
4. Utilização da função 'bisect', que isolará os índices que contêm as combinações válidas de 'y' dentro da lista de início ao fim.
5. Extração das somas através de um loop interno enxuto que passa apenas pelos índices válidos, adicionando o resultado no conjunto 'targets'.

```python
def compute_2sum(nums):
    targets = set()
    n = len(nums)

    for i in range(n):
        x = nums[i]
        low = -10000 - x
        high = 10000 - x

        left_idx = bisect.bisect_left(nums, low, i + 1, n)
        right_idx = bisect.bisect_right(nums, high, left_idx, n)

        for j in range(left_idx, right_idx):
            targets.add(x + nums[j])

    return len(targets)
```

## Execução e formatação da resposta
1. Execução da busca construída na célula anterior passando o array ordenado como entrada.
2. Formatação da resposta final extraindo o tamanho do conjunto de resultados, que representa exatamente o número de alvos distintos que puderam ser formados.

```python
final_answer = compute_2sum(nums)

print("==================================")
print("RESPOSTA FINAL:")
print(final_answer)
print("==================================")
```
